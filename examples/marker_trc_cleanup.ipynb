# Marker TRC Cleanup

Use this notebook when you already have a TRC file and want to inspect marker quality, fill short gaps, smooth trajectories, and write a cleaned TRC.

## 1. Imports and paths

```python
from pathlib import Path
import monomech as mm

TRC_PATH = Path("data/walk.trc")
OUTPUT_DIR = Path("outputs/walk")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

## 2. Load and summarize

```python
trial = mm.load_trc(TRC_PATH)

display(trial.summary().head(20))
print("Markers:", len(trial.marker_names))
print("Sampling rate:", trial.sampling_rate)
print("Time range:", trial.time_range)
```

## 3. Clean with the high-level API

```python
markers = mm.gap_fill(TRC_PATH, max_gap_frames=20)
markers = mm.smooth(markers, cutoff_hz=6.0)

display(markers.summary().head(20))
```

## 4. Preview and export

```python
markers.vis_3d(frame=50)
clean_trc = markers.to_trc(OUTPUT_DIR / "walk_clean.trc")
print(clean_trc)
```

## 5. Optional model marker check

```python
model_path = mm.get_builtin_osim_model("mocap")
marker_check = trial.validate_against_model(model_path)
display(marker_check.head(30))
print("Markers missing from model:", int((~marker_check["in_model"]).sum()))
```
