# Marker TRC Cleanup

Use this notebook when you already have a TRC file and want to inspect marker quality, fill short gaps, smooth trajectories, and write a cleaned TRC.

## 1. Import and set paths

```python
from pathlib import Path
import monomech as mm

TRC_PATH = Path("data/walk.trc")
OUTPUT_DIR = Path("outputs/walk")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("monomech import OK")
```

## 2. Load and summarize the trial

```python
trial = mm.load_trc(TRC_PATH)

print("Markers:", len(trial.marker_names))
print("First markers:", trial.marker_names[:10])
print("Sampling rate:", trial.sampling_rate)
print("Time range:", trial.time_range)

display(trial.summary().head(20))
```

## 3. Optional model marker check

If you already know the target OpenSim model, compare TRC marker names against that model before running IK.

```python
model_path = mm.get_builtin_osim_model("mocap")
marker_check = trial.validate_against_model(model_path)
display(marker_check.head(20))
```

## 4. Clean markers

Start with conservative settings. Increase `gap_fill_max_frames` only after checking which markers are missing and for how long.

```python
trial.clean_markers(
    gap_fill_method="rigid_cluster",
    gap_fill_max_frames=20,
    cutoff_hz=6.0,
)

display(trial.summary().head(20))
```

## 5. Export cleaned TRC

```python
clean_trc = trial.to_trc(OUTPUT_DIR / "walk_clean.trc")
print("Clean TRC:", clean_trc)
```
